# Sentiment Analysis with Hugging Face (Phân tích cảm xúc)

```python
# BƯỚC 1: Install 
# AutoTokenizer: cắt chữ thành số
# AutoModelForSequenceClassification :mô hình phân loại câu
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

# BƯỚC 2: Load model & tokenizer từ Hugging Face Hub
# Dùng model distilbert chuyên về cảm xúc cho Tiếng anh
model_name = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# BƯỚC 3: Tokenize một câu mẫu (Biến chữ thành số)
sample_sentence = "I'm so annoyed that it's raining today." #Thật tức dận vì hôm nay trời mưa
# return_tensors="pt" nghĩa là trả về định dạng cho PyTorch
inputs = tokenizer(sample_sentence, return_tensors="pt")

print("Kết quả Bước 3: Tokenization")
print(inputs) 

# BƯỚC 4: Perform Sentiment Analysis (Đưa số vào model để dự đoán)
with torch.no_grad():
    outputs = model(**inputs)

# Model sẽ tính toán và trả về điểm số 
# Lấy ra nhãn có điểm số cao nhất
predicted_class_id = outputs.logits.argmax().item()
ket_qua = model.config.id2label[predicted_class_id]

print("\n Kết quả Bước 4: Phân tích cảm xúc")
print("Câu văn:", sample_sentence)
print("Dự đoán:", ket_qua) # Kết quả sẽ là POSITIVE hoặc NEGATIVE
```
